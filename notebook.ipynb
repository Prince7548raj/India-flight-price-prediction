## Columns Distrbution

```python
num_cols = df.select_dtypes(exclude = "object").columns.drop("price")
```

## Feature & Target

```python
feature = df.drop(["price"], axis=1)
```

```python
target = df["price"]
```

## Train & Split

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(feature, target, test_size=0.2, random_state=42)
```

## Numeric Pipeline

```python
from sklearn.pipeline import Pipeline 
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
```

```python
num_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy= "median")),
    ("scaler", StandardScaler())
])
```

```python
cat_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("scaler", OneHotEncoder(handle_unknown="ignore"))
])
```

## Columns Transformer

```python
from sklearn.compose import ColumnTransformer

processor = ColumnTransformer([
    ("num", num_pipe, num_cols),
    ("cat", cat_pipe, cat_cols)
])
```

## Final Pipeline

```python
from sklearn.linear_model import LinearRegression

model = Pipeline([
    ("processor", processor),
    ("regressor", LinearRegression())
])
```

```python
model.fit(x_train, y_train)
```

## Test Data Prediction

```python
y_pred_test = model.predict(x_test)
y_pred_train = model.predict(x_train)
```

## Model Evaluation

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

r2_test = r2_score(y_test, y_pred_test)
r2_train = r2_score(y_train, y_pred_train)
mae = mean_absolute_error(y_test, y_pred_test)
mse = mean_squared_error(y_test, y_pred_test)
rmse =  np.sqrt(mse)

print(f"R2_Test :- {r2_test:.4}")
print(f"R2_Train :- {r2_train:.4}")
print(f"MAE :- {mae:.4}")
print(f"MSE :- {mse:.4}")
print(f"RMSE :- {rmse:.4}")
```

## Random Forest 

```python
from sklearn.ensemble import RandomForestRegressor

model_rf = Pipeline([
    ("processor", processor),
    ("regressor", RandomForestRegressor(
        n_estimators=100,
        max_depth=12,          # Tree ko gehra hone se rokega, speed 10x badhegi
        max_samples=0.5,        # Har tree 50% data hi dekhega (training super fast ho jayegi)
        n_jobs=-1,             # Saare CPU cores use karega
        random_state=42
    ))
])
```

```python
model_rf.fit(x_train, y_train)
```

```python
y_pred_test_rf = model_rf.predict(x_test)
y_pred_train_rf = model_rf.predict(x_train)

r2_test_rf = r2_score(y_test, y_pred_test_rf)
r2_train_rf = r2_score(y_train, y_pred_train_rf)
mae_rf = mean_absolute_error(y_test, y_pred_test_rf)
rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_test_rf))

print(f"R2_Test :- {r2_test_rf:.4}")
print(f"R2_Train :- {r2_train_rf:.4}")
print(f"MAE :- {mae_rf:.4}")
print(f"RMSE :- {rmse_rf:.4}")
```

## XGBoost Model

```python
from xgboost import XGBRegressor

model_xgb = Pipeline([
    ("processor", processor),
    ("regressor", XGBRegressor(n_estimators=300, 
                               n_jobs=-1, 
                               random_state=42))
])
```

```python
model_xgb.fit(x_train, y_train)
```

```python
y_pred_test_xgb = model_xgb.predict(x_test)
y_pred_train_xgb = model_xgb.predict(x_train)

r2_test_xgb = r2_score(y_test, y_pred_test_xgb)
r2_train_xgb = r2_score(y_train, y_pred_train_xgb)
mae_xgb = mean_absolute_error(y_test, y_pred_test_xgb)
rmse_xgb = np.sqrt(mean_squared_error(y_test, y_pred_test_xgb))

print(f"R2_Test :- {r2_test_xgb:.4}")
print(f"R2_Train :- {r2_train_xgb:.4}")
print(f"MAE :- {mae_xgb:.4}")
print(f"RMSE :- {rmse_xgb:.4}")
```

## Hyperparameter Tuning (XGBoost)

```python
from sklearn.model_selection import RandomizedSearchCV

param_grid = {
    "regressor__n_estimators": [100, 150, 200],
    "regressor__max_depth": [6, 8],
    "regressor__learning_rate": [0.05, 0.1],
    "regressor__subsample": [0.8, 1.0]
}

search = RandomizedSearchCV(
    estimator=model_xgb,
    param_distributions=param_grid,
    n_iter=4,
    cv=3,
    scoring="r2",
    random_state=42,
    verbose=1,
    n_jobs=1    # Windows environment mein parallel process hang hone se bachata hai
)
```

```python
search.fit(x_train, y_train)
```

```python
print("Best Params :-", search.best_params_)
print("Best CV R2 :-", search.best_score_)
```

```python
best_model = search.best_estimator_

y_pred_test_best = best_model.predict(x_test)
y_pred_train_best = best_model.predict(x_train)

r2_test_best = r2_score(y_test, y_pred_test_best)
r2_train_best = r2_score(y_train, y_pred_train_best)
mae_best = mean_absolute_error(y_test, y_pred_test_best)
rmse_best = np.sqrt(mean_squared_error(y_test, y_pred_test_best))

print(f"R2_Test :- {r2_test_best:.4}")
print(f"R2_Train :- {r2_train_best:.4}")
print(f"MAE :- {mae_best:.4}")
print(f"RMSE :- {rmse_best:.4}")
```

## Model Comparison

```python
comparison = pd.DataFrame({
    "Model": ["Linear Regression", "Random Forest", "XGBoost", "XGBoost (Tuned)"],
    "R2_Test": [r2_test, r2_test_rf, r2_test_xgb, r2_test_best],
    "MAE": [mae, mae_rf, mae_xgb, mae_best],
    "RMSE": [rmse, rmse_rf, rmse_xgb, rmse_best]
})
comparison.round(3)
```
